```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("/content/cartoon-pictures.jpg")
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
def k_means(image, k, max_iters=10):
    pixels = image.reshape((-1, 3))
    centroids = pixels[np.random.choice(pixels.shape[0], k, replace=False)]
    for _ in range(max_iters):
        distances = np.linalg.norm(pixels[:, np.newaxis, :] - centroids, axis=2)
        labels = np.argmin(distances, axis=1)
        for i in range(k):
            centroids[i] = np.mean(pixels[labels == i], axis=0)
    segmented_image = centroids[labels].reshape(image.shape)
    return segmented_image

def plt_images(img,result_image):
  figure_size = 15
  plt.figure(figsize=(figure_size,figure_size))
  plt.subplot(1,2,1),plt.imshow(img)
  plt.title('Original Image'), plt.xticks([]), plt.yticks([])
  plt.subplot(1,2,2)
  plt.imshow(result_image.astype(np.uint8))
  plt.title('Segmented Image when K = %i' % K), plt.xticks([]), plt.yticks([])
  plt.show()
```

```python
for i in range(3,11):
  K = i
  attempts=100
  result_image = k_means(img,K,attempts)
  plt_images(img,result_image)
```
